```python
import random 
import time

"""Dictionaries and lists"""
choices = ["rock", "paper", "scissors"]
game_stats = {
    "current_round": 1 ,
    "player_wins" : 0,
    "computer_wins" : 0,
    "game_over" : False,
}

animations = {
     "rock_vs_scissors": r"""
        .------------.
      .'   ___        '.                    .--.
     /    /   \    _    \                  (    )
    |             / \    |      \   /       '--'\      /
    |   _                |    -- SMASH! --       \    /
     \     ______       /       /   \             \  /
      '._            _.'                           ><
         '----------'                             /  \
                                             .--./    \
                                            (    )
                                             '--'


                       ROCK WINS!
""",
    "scissors_vs_rock": r"""
                                          .------------.
         .--.                           .'        ___   '.
        (    )                         /    _    /   \    \
      \  '--'                         |    / \             |
       \ /          \   /             |                _   |
        X         -- SMASH! --         \       ______     /
       / \          /   \               '._            _.'
      /  .--.                              '----------'
        (    )
         '--'


                       ROCK WINS!
""",

"paper_vs_rock": r"""
       .------------.
       |            |\
       |  --------  |_\
       |              |                    .------------.
       |  ----------  |                  .'   ___        '.
       |              |     \   /       /    /   \    _    \
       |  ----------  |   -- WRAP! --  |             / \    |
       |              |     /   \     |   _                |
       |  ----------  |                \     ______       /
       |              |                 '._            _.'
       '--------------'                    '----------'


                       PAPER WINS!
""",

"rock_vs_paper": r"""
                                          .------------.
                                          |            |\
                                          |  --------  |_\
        .------------.                    |              |
      .'   ___        '.                  |  ----------  |
     /    /   \    _    \      \   /       |              |
    |             / \    |  -- WRAP! --   |  ----------  |
    |   _                |    /   \       |              |
     \     ______       /                 |  ----------  |
      '._            _.'                  |              |
         '----------'                     '--------------'


                       PAPER WINS!
""",

"scissors_vs_paper": r"""
                                          .------------.
         .--.                             |            |\
        (    )                            |  --------  |_\
      \  '--'                             |              |
       \ /          \   /                 |  ----------  |
        X         -- SNIP! --             | - - - - - -  |
       / \          /   \                 |  ----------  |
      /  .--.                             |              |
        (    )                            |  ----------  |
         '--'                             |              |
                                          '--------------'


                     SCISSORS WIN!
""",

"paper_vs_scissors": r"""
       .------------.
       |            |\                       .--.
       |  --------  |_\                     (    )
       |              |                      '--'\      /
       |  ----------  |      \   /                \    /
       | - - - - - -  |    -- SNIP! --              \  /
       |  ----------  |      /   \                  ><
       |              |                            /  \
       |  ----------  |                       .--./    \
       |              |                      (    )
       '--------------'                       '--'


                     SCISSORS WIN!
""",

"rock_vs_rock": r"""
        .------------.                    .------------.
      .'   ___        '.                .'        ___   '.
     /    /   \    _    \              /    _    /   \    \
    |             / \    |    \ /    |    / \             |
    |   _                | -- BONK! --|                _   |
     \     ______       /     / \     \       ______     /
      '._            _.'               '._            _.'
         '----------'                     '----------'


                          TIE!
""",

"paper_vs_paper": r"""
       .------------.                     .------------.
       |            |\                    |            |\
       |  --------  |_\                   |  --------  |_\
       |              |                   |              |
       |  ----------  |                   |  ----------  |
       |              |      \   /        |              |
       |  ----------  |    -- FLAP! --    |  ----------  |
       |              |      /   \        |              |
       |  ----------  |                   |  ----------  |
       |              |                   |              |
       '--------------'                   '--------------'


                          TIE!
""",

"scissors_vs_scissors": r"""
         .--.                                .--.
        (    )                              (    )
      \  '--'                                '--'\      /
       \ /             \   /                      \    /
        X            -- CLINK! --                  \  /
       / \             /   \                        ><
      /  .--.                                      /  \
        (    )                                .--./    \
         '--'                                (    )
                                              '--'


                          TIE!
""",
     
}

"""Functions below"""

def reset_game(game_stats):
    game_stats["player_wins"] = 0
    game_stats["computer_wins"] = 0
    game_stats["current_round"] = 1
    game_stats["game_over"] = False
    


def game_state(winner):
    """count the current rounds"""
    if winner == "player":
        game_stats["player_wins"] += 1
    elif winner == "computer":
        game_stats["computer_wins"] += 1
    game_stats["current_round"] +=1
    return game_stats

def computer_choice():
    """make the computer choose one option"""
    computer_move = random.choice(choices)
    print(f"Computer uses {computer_move}")
    return computer_move

def player_choice():
    """validate player choosen option"""
    player_move = input("Choose your move!: ").lower().strip()
    while player_move not in choices and player_move!= "x":
        print("Wrong move choose between ROCK, PAPER, SCISSORS!")
        print("press X to close the game!")
        player_move = input("Choose your move!: ").lower().strip()
    if player_move == "x":
        game_stats["game_over"] = True
        return "exit"
    
        
    print(f"Player uses {player_move}")
    return player_move

def comparing_moves(Player, Ai):
    """this function compare player choices vs ai choices adn determine outcome"""
    
    if Ai == "rock" and Player == "rock": 
        print("It's a tie!")
        show_animation(animations["rock_vs_rock"])
        return "tie"
    elif Ai == "paper" and Player == "paper":
        print("It's a tie!")
        show_animation(animations["paper_vs_paper"])
        return "tie"
    elif Ai == "scissors" and Player == "scissors":
        print("It's a tie!")
        show_animation(animations["scissors_vs_scissors"])
        return "tie"
    elif Ai == "rock" and Player == "paper":
        print("Paper covers rock")
        print("Player wins")
        show_animation(animations["paper_vs_rock"])
        return "player"
    elif Ai == "paper" and Player == "scissors":
        print("Scissors cut paper")
        print("Player wins")
        show_animation(animations["scissors_vs_paper"])
        return "player"
    elif Ai == "scissors" and Player == "rock":
        print("Rock breaks scissors")
        print("Player wins")
        show_animation(animations["rock_vs_scissors"])        
        return "player"
    elif Ai == "paper" and Player == "rock":
        print("Paper covers rock")
        print("Computer wins")
        show_animation(animations["rock_vs_paper"])
        return "computer"
    elif Ai == "scissors" and Player == "paper":
        print("Scissors cut paper")
        print("Computer wins")
        show_animation(animations["paper_vs_scissors"])
        return "computer"
    elif Ai == "rock" and Player == "scissors":
        print("Rock breaks scissors")
        print("Computer wins")
        show_animation(animations["scissors_vs_rock"])
        return "computer"

def game_over():
    """this function count the wins of each player up to 3"""
    if game_stats["player_wins"] >= 3:
        print("The game is over")
        game_stats["game_over"] = True
        print("You are the winner!")
    elif game_stats["computer_wins"] >= 3:
        print("You are the loser")
        game_stats["game_over"] = True

def display_status():
    """Display the current round and scores."""
    print("\n--------------------")
    print(f"ROUND {game_stats['current_round']}")
    print(f"Player: {game_stats['player_wins']} | Computer: {game_stats['computer_wins']}")
    print("--------------------")

def display_menu():

    print("""
------ MAIN MENU ------

     1. SINGLE GAME
     2. TOURNAMENT
     3. EXIT

-----------------------
""")
    x = int(input)
    if x == 1:
        return "Single Game"
    elif x == 2:
        return "tournament"
    elif x == 3:
        print("BYE BYE")
        return "exit"
    
def welcome_title():
    print("""
    Hello
    welcome to my game
    you will be playing:
    ROCK,PAPER, SCISSORS
    CHOOSE A MOVE!
        
        
        
        """)

def single_game():
    reset_game(game_stats)
    while game_stats["game_over"] == False:
        display_status()
        player_move = player_choice()
        if player_move == "exit":
            break
        computer_move = computer_choice()
        winner = comparing_moves(player_move, computer_move)
        game_state(winner)
        game_over()
        if not game_stats["game_over"]:
            print("next round")

def tounament():
    loop = 0
    while loop < 2:
        display_status()
        player_move = player_choice()
        if player_move == "exit":
            break
        computer_move = computer_choice()
        winner = comparing_moves(player_move, computer_move)
        game_state(winner)
        game_over()
        if not game_stats["game_over"]:
            print("next round")






def show_animation(animations):
    """this functions prints an animation for player and ai moves"""
    for line in animations.splitlines():
        print(line, flush=True)
        time.sleep(0.15)
     



def play_game():
    """main function that runs the game"""
    welcome_title
    reset_game(game_stats)
    while game_stats["game_over"] == False:
        display_status()
        player_move = player_choice()
        if player_move == "exit":
            break
        computer_move = computer_choice()
        winner = comparing_moves(player_move, computer_move)
        game_state(winner)
        game_over()
        if not game_stats["game_over"]:
            print("next round")

    print(f"Your final score: {game_stats['player_wins']}")
    print(f"Computer final score: {game_stats['computer_wins']}")
    


         
```
